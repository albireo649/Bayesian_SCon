```python
%pip install ase
from ase.optimize import BFGS
import numpy as np
from ase.calculators.dftb import Dftb
from ase.io import read, write
import os
```

```python
# Read the molecule
atoms = read('../sys.xyz')

# box size
atoms.set_pbc([True, True, True])
atoms.set_cell([25,25,25])
atoms.center()
```

```python
calc = Dftb(label='sys',
	   #Hamiltonian_SlaterKosterFiles_Prefix = "/home/wonk/src/par/trans3d-0-1/",
	   Hamiltonian_SlaterKosterFiles_Prefix = "./corrupt_pp/",
	   Hamiltonian_MaxAngularMomentum_='',
	   Hamiltonian_MaxAngularMomentum_C='p',
	   Hamiltonian_MaxAngularMomentum_H='s',
	   kpts=(1,1,1)
	   )

atoms.set_calculator(calc)
```

```python
import shutil
import shlex

# Pre-optimize wigh BFGS optimization module
dftb_exe = shutil.which("dftb+")
if dftb_exe is None:
    raise RuntimeError("DFTB+ is not installed or dftb+ is not on PATH.")

calc.command = f"{shlex.quote(dftb_exe)} > PREFIX.out"

dyn = BFGS(atoms=atoms, trajectory="opt.traj")
dyn.run(fmax=0.01)

write("opt.cif", atoms)

write('opt.cif', atoms)
```
